### Penggunaan Sederhana - Grafik Pertama Anda

Mari kita mulai dengan contoh sederhana. Kita akan membuat scatter plot menggunakan dataset bawaan Plotly (iris dataset).

**Dataset Iris** adalah dataset klasik yang berisi pengukuran 150 bunga iris dari 3 spesies berbeda:
- Sepal length & width (panjang & lebar kelopak luar)
- Petal length & width (panjang & lebar mahkota)
- Species (spesies bunga)

````python
# Contoh penggunaan sederhana - Scatter Plot
# Load dataset iris (dataset bawaan Plotly)
df_iris = px.data.iris()

# Tampilkan 5 baris pertama untuk melihat struktur data
print("5 baris pertama dataset iris:")
print(df_iris.head())
print(f"\nJumlah data: {len(df_iris)} baris")
print(f"Kolom: {list(df_iris.columns)}")
````

````python
# Membuat scatter plot sederhana
# x-axis: panjang kelopak (sepal_length)
# y-axis: lebar kelopak (sepal_width)
# color: warna berdasarkan spesies
fig = px.scatter(
    df_iris, 
    x='sepal_length', 
    y='sepal_width',
    color='species',
    title='Scatter Plot: Sepal Length vs Sepal Width'
)

# Menampilkan grafik
fig.show()

# Penjelasan:
# - px.scatter() membuat scatter plot
# - parameter 'color' membedakan titik berdasarkan kategori (spesies)
# - Grafik bersifat INTERAKTIF: Anda bisa zoom, pan, dan hover untuk melihat detail
````

## 2. Basic Charts (Grafik Dasar)

Di bagian ini, kita akan mempelajari berbagai jenis grafik dasar yang sering digunakan:

### 2.1 Scatter Plot (Grafik Titik)

Scatter plot digunakan untuk menunjukkan hubungan antara dua variabel numerik. Setiap titik merepresentasikan satu observasi.

**Kapan menggunakan**: Untuk melihat korelasi, pola, atau outliers dalam data.

````python
# Scatter Plot dengan lebih banyak customization
df_iris = px.data.iris()

fig = px.scatter(
    df_iris,
    x='sepal_length',
    y='sepal_width',
    color='species',              # Warna berdasarkan spesies
    size='petal_length',          # Ukuran titik berdasarkan petal_length
    hover_data=['petal_width'],   # Data tambahan saat hover
    title='Scatter Plot: Hubungan Sepal Length & Width',
    labels={                       # Custom label untuk axis
        'sepal_length': 'Panjang Sepal (cm)',
        'sepal_width': 'Lebar Sepal (cm)',
        'species': 'Spesies'
    },
    size_max=15                    # Ukuran maksimum titik
)

# Menambahkan grid dan theme
fig.update_layout(
    plot_bgcolor='rgba(240,240,240,0.5)',
    font=dict(size=12)
)

fig.show()

# Insight: Kita bisa melihat pola clustering berdasarkan spesies
# dan hubungan antara ukuran sepal dengan petal
````

### 2.2 Line Chart (Grafik Garis)

Line chart digunakan untuk menunjukkan trend atau perubahan data sepanjang waktu.

**Kapan menggunakan**: Time series data, trend analysis, continuous data.

````python
# Line Chart - Data harga saham (dataset bawaan)
df_stocks = px.data.stocks()

print("Dataset harga saham:")
print(df_stocks.head())
print(f"\nKolom: {list(df_stocks.columns)}")
````

````python
# Membuat line chart untuk multiple saham
fig = px.line(
    df_stocks,
    x='date',
    y=['GOOG', 'AAPL', 'AMZN', 'FB', 'NFLX', 'MSFT'],
    title='Pergerakan Harga Saham Tech Giants (2018-2020)',
    labels={
        'date': 'Tanggal',
        'value': 'Harga Saham (USD)',
        'variable': 'Perusahaan'
    }
)

# Customization layout
fig.update_layout(
    hovermode='x unified',  # Hover menampilkan semua nilai pada x yang sama
    legend=dict(
        orientation="h",     # Legend horizontal
        yanchor="bottom",
        y=1.02,
        xanchor="right",
        x=1
    )
)

fig.show()

# Penjelasan: Grafik ini menunjukkan pergerakan harga 6 saham teknologi besar
# Kita bisa membandingkan trend dan volatilitas masing-masing saham
````

### 2.3 Bar Chart (Grafik Batang)

Bar chart digunakan untuk membandingkan nilai antar kategori.

**Kapan menggunakan**: Membandingkan jumlah, frekuensi, atau nilai antar grup/kategori.

````python
# Bar Chart - Data tips restoran
df_tips = px.data.tips()

print("Dataset tips restoran:")
print(df_tips.head(10))
print(f"\nTotal data: {len(df_tips)} transaksi")
````

````python
# Vertical Bar Chart - Total tip berdasarkan hari
total_tips_by_day = df_tips.groupby('day')['tip'].sum().reset_index()

fig = px.bar(
    total_tips_by_day,
    x='day',
    y='tip',
    title='Total Tips per Hari',
    labels={'day': 'Hari', 'tip': 'Total Tips ($)'},
    color='tip',                    # Warna berdasarkan nilai
    color_continuous_scale='Blues',  # Skala warna
    text='tip'                       # Tampilkan nilai di bar
)

# Format text pada bar
fig.update_traces(texttemplate='$%{text:.2f}', textposition='outside')

fig.show()

# Insight: Kita bisa melihat hari mana yang menghasilkan tips terbanyak
````

````python
# Grouped Bar Chart - Membandingkan tips berdasarkan gender dan hari
tips_gender_day = df_tips.groupby(['day', 'sex'])['tip'].mean().reset_index()

fig = px.bar(
    tips_gender_day,
    x='day',
    y='tip',
    color='sex',  # Grouping berdasarkan gender
    barmode='group',  # Mode: 'group', 'stack', atau 'overlay'
    title='Rata-rata Tips: Perbedaan Gender per Hari',
    labels={'day': 'Hari', 'tip': 'Rata-rata Tip ($)', 'sex': 'Gender'},
    text='tip'
)

fig.update_traces(texttemplate='$%{text:.2f}', textposition='outside')

fig.show()

# Penjelasan: Grouped bar chart memudahkan perbandingan antar sub-kategori
# Kita bisa melihat perbedaan tips antara pria dan wanita di setiap hari
````

### 2.4 Pie Chart & Histogram

**Pie Chart** menunjukkan proporsi atau persentase dari keseluruhan.  
**Histogram** menunjukkan distribusi frekuensi dari data numerik.

Mari kita lihat contoh keduanya:

````python
# Pie Chart - Distribusi tips berdasarkan hari
tips_by_day = df_tips.groupby('day')['tip'].sum().reset_index()

fig = px.pie(
    tips_by_day,
    values='tip',
    names='day',
    title='Distribusi Tips Berdasarkan Hari',
    hole=0,  # 0 = pie chart biasa, 0.3-0.5 = donut chart
    color_discrete_sequence=px.colors.sequential.RdBu  # Color scheme
)

# Tambahkan persentase pada label
fig.update_traces(
    textposition='inside',
    textinfo='percent+label'
)

fig.show()
````

````python
# Histogram - Distribusi total bill
fig = px.histogram(
    df_tips,
    x='total_bill',
    nbins=20,  # Jumlah bins (kolom)
    title='Distribusi Total Bill di Restoran',
    labels={'total_bill': 'Total Bill ($)'},
    color_discrete_sequence=['indianred']
)

# Tambahkan garis rata-rata
mean_bill = df_tips['total_bill'].mean()
fig.add_vline(
    x=mean_bill,
    line_dash="dash",
    line_color="green",
    annotation_text=f"Mean: ${mean_bill:.2f}",
    annotation_position="top right"
)

fig.show()

# Insight: Histogram menunjukkan bahwa kebanyakan bill berada di range $10-$25
# Distribusi right-skewed (ekor ke kanan)
````

### 2.5 Box Plot (Grafik Kotak)

Box plot menunjukkan distribusi data melalui quartiles, median, dan outliers.

**Komponen Box Plot:**
- **Box**: Q1 (25%) hingga Q3 (75%) - Interquartile Range (IQR)
- **Line dalam box**: Median (Q2, 50%)
- **Whiskers**: Min dan Max (dalam 1.5 * IQR)
- **Points**: Outliers

````python
# Box Plot - Distribusi tips berdasarkan hari
fig = px.box(
    df_tips,
    x='day',
    y='tip',
    color='time',  # Warna berdasarkan waktu makan
    title='Distribusi Tips Berdasarkan Hari dan Waktu',
    labels={'day': 'Hari', 'tip': 'Tip ($)', 'time': 'Waktu'},
    points='all'  # Tampilkan semua data points
)

fig.show()

# Insight: Box plot membantu kita melihat:
# 1. Median tips di setiap hari
# 2. Variabilitas (IQR) tips
# 3. Outliers (tips yang sangat tinggi/rendah)
# 4. Perbandingan antara lunch dan dinner
````

## 3. Statistical Charts (Grafik Statistik)

Di bagian ini kita akan belajar grafik-grafik untuk analisis statistik yang lebih advanced:

### 3.1 Violin Plot

Violin plot menggabungkan box plot dengan kernel density plot, menunjukkan distribusi data dengan lebih detail.

**Keunggulan**: Menunjukkan distribusi lengkap (tidak hanya quartiles), termasuk multimodal distribution.

````python
# Violin Plot - Distribusi tips yang lebih detail
fig = px.violin(
    df_tips,
    y='tip',
    x='day',
    color='sex',
    box=True,  # Tampilkan box plot di dalam violin
    points='all',  # Tampilkan data points
    title='Violin Plot: Distribusi Tips Berdasarkan Hari dan Gender',
    labels={'day': 'Hari', 'tip': 'Tip ($)', 'sex': 'Gender'}
)

fig.show()

# Penjelasan: 
# - Lebar violin menunjukkan density (banyaknya data di nilai tersebut)
# - Box plot di dalam menunjukkan quartiles
# - Points menunjukkan data aktual
````

### 3.2 Scatter Matrix (Pair Plot)

Scatter matrix menampilkan scatter plots untuk setiap pasangan variabel dalam dataset, berguna untuk exploratory data analysis (EDA).

**Kapan menggunakan**: Untuk melihat hubungan antar multiple variabel sekaligus.

````python
# Scatter Matrix - Iris dataset
fig = px.scatter_matrix(
    df_iris,
    dimensions=['sepal_length', 'sepal_width', 'petal_length', 'petal_width'],
    color='species',
    title='Scatter Matrix: Iris Dataset',
    labels={col: col.replace('_', ' ').title() for col in df_iris.columns},
    height=800
)

# Update diagonal untuk histogram
fig.update_traces(diagonal_visible=False)

fig.show()

# Insight: Scatter matrix membantu kita:
# 1. Melihat korelasi antar variabel
# 2. Mengidentifikasi variabel yang memisahkan spesies dengan baik
# 3. Menemukan pola dan outliers
````

### 3.3 Correlation Heatmap

Heatmap menunjukkan korelasi antar variabel numerik menggunakan warna.

**Kapan menggunakan**: Untuk melihat hubungan linear antar banyak variabel, feature selection.

````python
# Heatmap - Correlation matrix
# Hitung korelasi antar variabel numerik
correlation_matrix = df_iris[['sepal_length', 'sepal_width', 'petal_length', 'petal_width']].corr()

# Membuat heatmap dengan graph_objects
fig = go.Figure(data=go.Heatmap(
    z=correlation_matrix.values,
    x=correlation_matrix.columns,
    y=correlation_matrix.columns,
    colorscale='RdBu',  # Red-Blue color scale
    zmid=0,  # Center color pada 0
    text=correlation_matrix.values,
    texttemplate='%{text:.2f}',
    textfont={"size": 12},
    colorbar=dict(title="Correlation")
))

fig.update_layout(
    title='Correlation Heatmap: Iris Dataset',
    xaxis_title='Features',
    yaxis_title='Features',
    width=600,
    height=600
)

fig.show()

# Interpretasi:
# - Nilai mendekati 1 (merah): Korelasi positif kuat
# - Nilai mendekati -1 (biru): Korelasi negatif kuat
# - Nilai mendekati 0 (putih): Tidak ada korelasi
````

### 3.4 3D Scatter Plot

Scatter plot 3 dimensi untuk visualisasi data dengan 3 variabel numerik.

**Kapan menggunakan**: Analisis data 3D, clustering visualization, multivariate analysis.

````python
# 3D Scatter Plot - Iris dataset
fig = px.scatter_3d(
    df_iris,
    x='sepal_length',
    y='sepal_width',
    z='petal_length',
    color='species',
    size='petal_width',
    title='3D Scatter Plot: Iris Dataset',
    labels={
        'sepal_length': 'Sepal Length (cm)',
        'sepal_width': 'Sepal Width (cm)',
        'petal_length': 'Petal Length (cm)',
        'species': 'Species'
    },
    size_max=15,
    opacity=0.7
)

# Customization camera angle
fig.update_layout(
    scene=dict(
        xaxis_title='Sepal Length',
        yaxis_title='Sepal Width',
        zaxis_title='Petal Length',
    )
)

fig.show()

# Fitur Interaktif 3D:
# - Rotate: Click dan drag untuk memutar
# - Zoom: Scroll atau pinch
# - Pan: Shift + click dan drag
````

## 4. Dash - Framework Web untuk Plotly

**Dash** adalah framework Python untuk membangun web application dengan visualisasi data interaktif. Dash menggunakan Plotly untuk grafik dan React.js untuk UI, tanpa perlu menulis JavaScript.

**Kegunaan Dash:**
- Membuat dashboard interaktif
- Web application untuk data analytics
- Real-time monitoring systems
- Report dan presentation tools

### 4.1 Instalasi dan Struktur Dasar Dash

Dash terdiri dari beberapa komponen:
- **dash**: Core framework
- **dash-core-components**: Komponen interaktif (grafik, dropdown, slider, dll)
- **dash-html-components**: Komponen HTML
- **dash-bootstrap-components**: Bootstrap styling (opsional)

Dash app terdiri dari 2 komponen utama:
1. **Layout**: Struktur visual app (HTML + components)
2. **Callbacks**: Interaktivitas (menghubungkan input dan output)

````python
# Instalasi Dash
# Jalankan perintah ini di terminal:
# pip install dash
# pip install dash-bootstrap-components  # (opsional, untuk styling)

print("Untuk menggunakan Dash, install terlebih dahulu:")
print("pip install dash dash-bootstrap-components")
````

### 4.2 Contoh Dash App Sederhana

**Catatan**: Contoh Dash di bawah ini adalah kode untuk file Python terpisah (.py), karena Dash memerlukan server web untuk berjalan.

Simpan kode berikut dalam file `app_simple.py` dan jalankan dengan: `python app_simple.py`

```python
import dash
from dash import dcc, html
import plotly.express as px

# Inisialisasi Dash app
app = dash.Dash(__name__)

# Load data
df = px.data.gapminder()

# Buat grafik
fig = px.scatter(
    df[df['year'] == 2007],
    x='gdpPercap',
    y='lifeExp',
    size='pop',
    color='continent',
    hover_name='country',
    title='Life Expectancy vs GDP (2007)',
    log_x=True
)

# Layout app
app.layout = html.Div([
    html.H1('Dashboard Sederhana dengan Dash'),
    html.P('Ini adalah contoh dashboard sederhana'),
    dcc.Graph(id='scatter-plot', figure=fig)
])

# Jalankan server
if __name__ == '__main__':
    app.run_server(debug=True)
```

**Penjelasan:**
- `app = dash.Dash(__name__)`: Inisialisasi aplikasi
- `app.layout`: Mendefinisikan struktur visual (HTML)
- `dcc.Graph`: Komponen untuk menampilkan grafik Plotly
- `app.run_server()`: Menjalankan web server (akses di http://127.0.0.1:8050/)

### 4.3 Dash App dengan Interaktivity (Callbacks)

Callback membuat app interaktif dengan menghubungkan input (dropdown, slider, dll) ke output (grafik, text, dll).

**Struktur Callback:**
```python
@app.callback(
    Output(component_id, component_property),
    Input(component_id, component_property)
)
def update_function(input_value):
    # Process input
    return output_value
```

**Contoh Lengkap dengan Callbacks** - Simpan dalam `app_interactive.py`:

```python
import dash
from dash import dcc, html, Input, Output
import plotly.express as px

app = dash.Dash(__name__)
df = px.data.gapminder()

# Layout
app.layout = html.Div([
    html.H1('Dashboard Interaktif - Gapminder Dataset'),
    
    html.Div([
        html.Label('Pilih Tahun:'),
        dcc.Slider(
            id='year-slider',
            min=df['year'].min(),
            max=df['year'].max(),
            step=5,
            value=2007,
            marks={str(year): str(year) for year in df['year'].unique()}
        )
    ]),
    
    dcc.Graph(id='scatter-plot')
])

# Callback untuk update grafik
@app.callback(
    Output('scatter-plot', 'figure'),
    Input('year-slider', 'value')
)
def update_graph(selected_year):
    filtered_df = df[df['year'] == selected_year]
    
    fig = px.scatter(
        filtered_df,
        x='gdpPercap',
        y='lifeExp',
        size='pop',
        color='continent',
        hover_name='country',
        title=f'Life Expectancy vs GDP ({selected_year})',
        log_x=True
    )
    
    return fig

if __name__ == '__main__':
    app.run_server(debug=True)
```

**Penjelasan:**
- Slider mengontrol tahun yang ditampilkan
- Saat slider berubah, callback `update_graph()` dipanggil
- Grafik otomatis update dengan data tahun yang dipilih
- Interaktif dan real-time!

### 4.4 Tips dan Best Practices Dash

**1. Layout Organization:**
- Gunakan `dbc.Container`, `dbc.Row`, `dbc.Col` untuk layout responsive
- Pisahkan styling dengan CSS external atau inline styles
- Gunakan `className` untuk Bootstrap classes

**2. Callbacks Optimization:**
- Batasi jumlah callbacks untuk performa optimal
- Combine multiple outputs dalam satu callback (lebih efisien)
- Gunakan `prevent_initial_call=True` jika tidak perlu render awal

**3. Komponen Dash yang Sering Digunakan:**

**Core Components (dcc):**
- `dcc.Graph`: Untuk Plotly charts
- `dcc.Dropdown`: Dropdown menu
- `dcc.Slider`: Slider untuk range
- `dcc.DatePickerRange`: Date picker
- `dcc.Input`: Text input
- `dcc.Loading`: Loading spinner

**HTML Components:**
- `html.Div`: Container
- `html.H1`, `html.H2`: Headers
- `html.P`: Paragraph
- `html.Button`: Button

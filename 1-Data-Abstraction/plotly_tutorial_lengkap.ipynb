`````python
````xml
<VSCode.Cell language="markdown">
# Tutorial Plotly - Visualisasi Data Interaktif

Selamat datang di tutorial lengkap Plotly! Tutorial ini akan membahas:
1. **Instalasi dan Penggunaan Sederhana**
2. **Basic Charts** (Grafik Dasar)
3. **Statistical Charts** (Grafik Statistik)
4. **Dash** (Framework Web untuk Plotly)

Plotly adalah library visualisasi data yang powerful untuk membuat grafik interaktif dengan Python, R, dan JavaScript. Plotly sangat cocok untuk analisis data, dashboard interaktif, dan presentasi data yang menarik.
</VSCode.Cell>

<VSCode.Cell language="markdown">
---
## 1. Instalasi dan Penggunaan Sederhana

### Instalasi Plotly

Untuk menginstal Plotly, gunakan pip (Python package manager). Ada beberapa package yang perlu diinstall:
</VSCode.Cell>

<VSCode.Cell language="python">
# Instalasi Plotly menggunakan pip
# Jalankan perintah ini di terminal atau command prompt:
# pip install plotly
# pip install plotly-express  # (opsional, sudah termasuk dalam plotly versi terbaru)
# pip install kaleido  # untuk export gambar statis

# Untuk tutorial Dash (akan dibahas di bagian 4):
# pip install dash
</VSCode.Cell>

<VSCode.Cell language="markdown">
### Import Library

Setelah instalasi selesai, kita perlu mengimport library yang diperlukan:
- **plotly.express**: API tingkat tinggi untuk membuat grafik dengan cepat
- **plotly.graph_objects**: API tingkat rendah untuk kontrol yang lebih detail
- **pandas**: Untuk manipulasi data
- **numpy**: Untuk operasi numerik
</VSCode.Cell>

<VSCode.Cell language="python">
# Import library yang diperlukan
import plotly.express as px
import plotly.graph_objects as go
import pandas as pd
import numpy as np

print("Plotly version:", px.__version__)
print("Instalasi berhasil! Siap membuat visualisasi interaktif.")
</VSCode.Cell>

<VSCode.Cell language="markdown">
### Penggunaan Sederhana - Grafik Pertama Anda

Mari kita mulai dengan contoh sederhana. Kita akan membuat scatter plot menggunakan dataset bawaan Plotly (iris dataset).

**Dataset Iris** adalah dataset klasik yang berisi pengukuran 150 bunga iris dari 3 spesies berbeda:
- Sepal length & width (panjang & lebar kelopak luar)
- Petal length & width (panjang & lebar mahkota)
- Species (spesies bunga)
</VSCode.Cell>

<VSCode.Cell language="python">
# Contoh penggunaan sederhana - Scatter Plot
# Load dataset iris (dataset bawaan Plotly)
df_iris = px.data.iris()

# Tampilkan 5 baris pertama untuk melihat struktur data
print("5 baris pertama dataset iris:")
print(df_iris.head())
print(f"\nJumlah data: {len(df_iris)} baris")
print(f"Kolom: {list(df_iris.columns)}")
</VSCode.Cell>

<VSCode.Cell language="python">
# Membuat scatter plot sederhana
# x-axis: panjang kelopak (sepal_length)
# y-axis: lebar kelopak (sepal_width)
# color: warna berdasarkan spesies
fig = px.scatter(
    df_iris, 
    x='sepal_length', 
    y='sepal_width',
    color='species',
    title='Scatter Plot: Sepal Length vs Sepal Width'
)

# Menampilkan grafik
fig.show()

# Penjelasan:
# - px.scatter() membuat scatter plot
# - parameter 'color' membedakan titik berdasarkan kategori (spesies)
# - Grafik bersifat INTERAKTIF: Anda bisa zoom, pan, dan hover untuk melihat detail
</VSCode.Cell>

<VSCode.Cell language="markdown">
**Fitur Interaktif Plotly:**
- 🖱️ **Hover**: Arahkan mouse ke titik untuk melihat detail data
- 🔍 **Zoom**: Klik dan drag untuk zoom in, double-click untuk reset
- 📸 **Save**: Tombol kamera di toolbar untuk save gambar
- 👁️ **Legend**: Klik legend untuk show/hide kategori
- ➕ **Pan**: Geser grafik dengan drag
</VSCode.Cell>

<VSCode.Cell language="markdown">
---
## 2. Basic Charts (Grafik Dasar)

Di bagian ini, kita akan mempelajari berbagai jenis grafik dasar yang sering digunakan:
- Scatter Plot (Grafik Titik)
- Line Chart (Grafik Garis)
- Bar Chart (Grafik Batang)
- Pie Chart (Grafik Lingkaran)
- Histogram (Grafik Distribusi)
- Box Plot (Grafik Kotak)
</VSCode.Cell>

<VSCode.Cell language="markdown">
### 2.1 Scatter Plot (Grafik Titik)

Scatter plot digunakan untuk menunjukkan hubungan antara dua variabel numerik. Setiap titik merepresentasikan satu observasi.

**Kapan menggunakan**: Untuk melihat korelasi, pola, atau outliers dalam data.
</VSCode.Cell>

<VSCode.Cell language="python">
# Scatter Plot dengan lebih banyak customization
df_iris = px.data.iris()

fig = px.scatter(
    df_iris,
    x='sepal_length',
    y='sepal_width',
    color='species',              # Warna berdasarkan spesies
    size='petal_length',          # Ukuran titik berdasarkan petal_length
    hover_data=['petal_width'],   # Data tambahan saat hover
    title='Scatter Plot: Hubungan Sepal Length & Width',
    labels={                       # Custom label untuk axis
        'sepal_length': 'Panjang Sepal (cm)',
        'sepal_width': 'Lebar Sepal (cm)',
        'species': 'Spesies'
    },
    size_max=15                    # Ukuran maksimum titik
)

# Menambahkan grid dan theme
fig.update_layout(
    plot_bgcolor='rgba(240,240,240,0.5)',
    font=dict(size=12)
)

fig.show()

# Insight: Kita bisa melihat pola clustering berdasarkan spesies
# dan hubungan antara ukuran sepal dengan petal
</VSCode.Cell>

<VSCode.Cell language="markdown">
### 2.2 Line Chart (Grafik Garis)

Line chart digunakan untuk menunjukkan trend atau perubahan data sepanjang waktu.

**Kapan menggunakan**: Time series data, trend analysis, continuous data.
</VSCode.Cell>

<VSCode.Cell language="python">
# Line Chart - Data harga saham Apple (dataset bawaan)
df_stocks = px.data.stocks()

print("Dataset harga saham:")
print(df_stocks.head())
print(f"\nKolom: {list(df_stocks.columns)}")
</VSCode.Cell>

<VSCode.Cell language="python">
# Membuat line chart untuk multiple saham
fig = px.line(
    df_stocks,
    x='date',
    y=['GOOG', 'AAPL', 'AMZN', 'FB', 'NFLX', 'MSFT'],
    title='Pergerakan Harga Saham Tech Giants (2018-2020)',
    labels={
        'date': 'Tanggal',
        'value': 'Harga Saham (USD)',
        'variable': 'Perusahaan'
    }
)

# Customization layout
fig.update_layout(
    hovermode='x unified',  # Hover menampilkan semua nilai pada x yang sama
    legend=dict(
        orientation="h",     # Legend horizontal
        yanchor="bottom",
        y=1.02,
        xanchor="right",
        x=1
    )
)

fig.show()

# Penjelasan: Grafik ini menunjukkan pergerakan harga 6 saham teknologi besar
# Kita bisa membandingkan trend dan volatilitas masing-masing saham
</VSCode.Cell>

<VSCode.Cell language="markdown">
### 2.3 Bar Chart (Grafik Batang)

Bar chart digunakan untuk membandingkan nilai antar kategori.

**Kapan menggunakan**: Membandingkan jumlah, frekuensi, atau nilai antar grup/kategori.
</VSCode.Cell>

<VSCode.Cell language="python">
# Bar Chart - Data tips restoran
df_tips = px.data.tips()

print("Dataset tips restoran:")
print(df_tips.head(10))
print(f"\nTotal data: {len(df_tips)} transaksi")
</VSCode.Cell>

<VSCode.Cell language="python">
# Vertical Bar Chart - Total tip berdasarkan hari
total_tips_by_day = df_tips.groupby('day')['tip'].sum().reset_index()

fig = px.bar(
    total_tips_by_day,
    x='day',
    y='tip',
    title='Total Tips per Hari',
    labels={'day': 'Hari', 'tip': 'Total Tips ($)'},
    color='tip',                    # Warna berdasarkan nilai
    color_continuous_scale='Blues',  # Skala warna
    text='tip'                       # Tampilkan nilai di bar
)

# Format text pada bar
fig.update_traces(texttemplate='$%{text:.2f}', textposition='outside')

fig.show()

# Insight: Kita bisa melihat hari mana yang menghasilkan tips terbanyak
</VSCode.Cell>

<VSCode.Cell language="python">
# Grouped Bar Chart - Membandingkan tips berdasarkan gender dan hari
tips_gender_day = df_tips.groupby(['day', 'sex'])['tip'].mean().reset_index()

fig = px.bar(
    tips_gender_day,
    x='day',
    y='tip',
    color='sex',  # Grouping berdasarkan gender
    barmode='group',  # Mode: 'group', 'stack', atau 'overlay'
    title='Rata-rata Tips: Perbedaan Gender per Hari',
    labels={'day': 'Hari', 'tip': 'Rata-rata Tip ($)', 'sex': 'Gender'},
    text='tip'
)

fig.update_traces(texttemplate='$%{text:.2f}', textposition='outside')

fig.show()

# Penjelasan: Grouped bar chart memudahkan perbandingan antar sub-kategori
# Kita bisa melihat perbedaan tips antara pria dan wanita di setiap hari
</VSCode.Cell>

<VSCode.Cell language="markdown">
### 2.4 Pie Chart (Grafik Lingkaran)

Pie chart menunjukkan proporsi atau persentase dari keseluruhan.

**Kapan menggunakan**: Menunjukkan komposisi atau distribusi kategori (max 5-7 kategori).
</VSCode.Cell>

<VSCode.Cell language="python">
# Pie Chart - Distribusi tips berdasarkan hari
tips_by_day = df_tips.groupby('day')['tip'].sum().reset_index()

fig = px.pie(
    tips_by_day,
    values='tip',
    names='day',
    title='Distribusi Tips Berdasarkan Hari',
    hole=0,  # 0 = pie chart biasa, 0.3-0.5 = donut chart
    color_discrete_sequence=px.colors.sequential.RdBu  # Color scheme
)

# Tambahkan persentase pada label
fig.update_traces(
    textposition='inside',
    textinfo='percent+label'
)

fig.show()

# Penjelasan: Pie chart menunjukkan proporsi tips di setiap hari
# Mudah melihat hari mana yang memberikan kontribusi terbesar
</VSCode.Cell>

<VSCode.Cell language="python">
# Donut Chart - Pie chart dengan hole di tengah
fig = px.pie(
    tips_by_day,
    values='tip',
    names='day',
    title='Distribusi Tips (Donut Chart)',
    hole=0.4  # Membuat hole di tengah (donut)
)

fig.update_traces(
    textposition='outside',
    textinfo='percent+label+value',
    texttemplate='%{label}<br>%{percent}<br>$%{value:.2f}'
)

fig.show()
</VSCode.Cell>

<VSCode.Cell language="markdown">
### 2.5 Histogram

Histogram menunjukkan distribusi frekuensi dari data numerik.

**Kapan menggunakan**: Untuk melihat distribusi, skewness, atau outliers dalam data.
</VSCode.Cell>

<VSCode.Cell language="python">
# Histogram - Distribusi total bill
fig = px.histogram(
    df_tips,
    x='total_bill',
    nbins=20,  # Jumlah bins (kolom)
    title='Distribusi Total Bill di Restoran',
    labels={'total_bill': 'Total Bill ($)'},
    color_discrete_sequence=['indianred']
)

# Tambahkan garis rata-rata
mean_bill = df_tips['total_bill'].mean()
fig.add_vline(
    x=mean_bill,
    line_dash="dash",
    line_color="green",
    annotation_text=f"Mean: ${mean_bill:.2f}",
    annotation_position="top right"
)

fig.show()

# Insight: Histogram menunjukkan bahwa kebanyakan bill berada di range $10-$25
# Distribusi right-skewed (ekor ke kanan) - ada beberapa bill yang sangat tinggi
</VSCode.Cell>

<VSCode.Cell language="python">
# Histogram dengan multiple groups
fig = px.histogram(
    df_tips,
    x='total_bill',
    color='time',  # Pisahkan berdasarkan waktu makan
    barmode='overlay',  # Mode: 'overlay', 'group', atau 'stack'
    nbins=15,
    title='Distribusi Total Bill: Lunch vs Dinner',
    labels={'total_bill': 'Total Bill ($)', 'time': 'Waktu'},
    opacity=0.7  # Transparansi untuk overlay
)

fig.show()

# Penjelasan: Kita bisa membandingkan distribusi bill antara lunch dan dinner
# Dinner cenderung memiliki bill yang lebih tinggi
</VSCode.Cell>

<VSCode.Cell language="markdown">
### 2.6 Box Plot (Grafik Kotak)

Box plot menunjukkan distribusi data melalui quartiles, median, dan outliers.

**Kapan menggunakan**: Untuk melihat spread, median, dan outliers dalam data.

**Komponen Box Plot:**
- **Box**: Q1 (25%) hingga Q3 (75%) - Interquartile Range (IQR)
- **Line dalam box**: Median (Q2, 50%)
- **Whiskers**: Min dan Max (dalam 1.5 * IQR)
- **Points**: Outliers
</VSCode.Cell>

<VSCode.Cell language="python">
# Box Plot - Distribusi tips berdasarkan hari
fig = px.box(
    df_tips,
    x='day',
    y='tip',
    color='time',  # Warna berdasarkan waktu makan
    title='Distribusi Tips Berdasarkan Hari dan Waktu',
    labels={'day': 'Hari', 'tip': 'Tip ($)', 'time': 'Waktu'},
    points='all'  # Tampilkan semua data points: 'all', 'outliers', False
)

fig.show()

# Insight: Box plot membantu kita melihat:
# 1. Median tips di setiap hari
# 2. Variabilitas (IQR) tips
# 3. Outliers (tips yang sangat tinggi/rendah)
# 4. Perbandingan antara lunch dan dinner
</VSCode.Cell>

<VSCode.Cell language="markdown">
---
## 3. Statistical Charts (Grafik Statistik)

Di bagian ini kita akan belajar grafik-grafik untuk analisis statistik yang lebih advanced:
- Violin Plot
- Scatter Matrix
- Correlation Heatmap
- Distribution Plot dengan Multiple Variables
- Contour Plot
- 3D Scatter Plot
</VSCode.Cell>

<VSCode.Cell language="markdown">
### 3.1 Violin Plot

Violin plot menggabungkan box plot dengan kernel density plot, menunjukkan distribusi data dengan lebih detail.

**Keunggulan**: Menunjukkan distribusi lengkap (tidak hanya quartiles), termasuk multimodal distribution.
</VSCode.Cell>

<VSCode.Cell language="python">
# Violin Plot - Distribusi tips yang lebih detail
fig = px.violin(
    df_tips,
    y='tip',
    x='day',
    color='sex',
    box=True,  # Tampilkan box plot di dalam violin
    points='all',  # Tampilkan data points
    title='Violin Plot: Distribusi Tips Berdasarkan Hari dan Gender',
    labels={'day': 'Hari', 'tip': 'Tip ($)', 'sex': 'Gender'}
)

fig.show()

# Penjelasan: 
# - Lebar violin menunjukkan density (banyaknya data di nilai tersebut)
# - Box plot di dalam menunjukkan quartiles
# - Points menunjukkan data aktual
# Violin plot lebih informatif daripada box plot karena menunjukkan shape distribusi
</VSCode.Cell>

<VSCode.Cell language="markdown">
### 3.2 Scatter Matrix (Pair Plot)

Scatter matrix menampilkan scatter plots untuk setiap pasangan variabel dalam dataset, berguna untuk exploratory data analysis (EDA).

**Kapan menggunakan**: Untuk melihat hubungan antar multiple variabel sekaligus.
</VSCode.Cell>

<VSCode.Cell language="python">
# Scatter Matrix - Iris dataset
fig = px.scatter_matrix(
    df_iris,
    dimensions=['sepal_length', 'sepal_width', 'petal_length', 'petal_width'],
    color='species',
    title='Scatter Matrix: Iris Dataset',
    labels={col: col.replace('_', ' ').title() for col in df_iris.columns},
    height=800
)

# Update diagonal untuk histogram
fig.update_traces(diagonal_visible=False)

fig.show()

# Insight: Scatter matrix membantu kita:
# 1. Melihat korelasi antar variabel
# 2. Mengidentifikasi variabel yang memisahkan spesies dengan baik
# 3. Menemukan pola dan outliers
# Contoh: petal_length & petal_width sangat baik memisahkan spesies
</VSCode.Cell>

<VSCode.Cell language="markdown">
### 3.3 Correlation Heatmap

Heatmap menunjukkan korelasi antar variabel numerik menggunakan warna.

**Kapan menggunakan**: Untuk melihat hubungan linear antar banyak variabel, feature selection.
</VSCode.Cell>

<VSCode.Cell language="python">
# Heatmap - Correlation matrix
# Hitung korelasi antar variabel numerik
correlation_matrix = df_iris[['sepal_length', 'sepal_width', 'petal_length', 'petal_width']].corr()

print("Correlation Matrix:")
print(correlation_matrix)
</VSCode.Cell>

<VSCode.Cell language="python">
# Membuat heatmap dengan graph_objects untuk lebih custom
fig = go.Figure(data=go.Heatmap(
    z=correlation_matrix.values,
    x=correlation_matrix.columns,
    y=correlation_matrix.columns,
    colorscale='RdBu',  # Red-Blue color scale
    zmid=0,  # Center color pada 0
    text=correlation_matrix.values,
    texttemplate='%{text:.2f}',
    textfont={"size": 12},
    colorbar=dict(title="Correlation")
))

fig.update_layout(
    title='Correlation Heatmap: Iris Dataset',
    xaxis_title='Features',
    yaxis_title='Features',
    width=600,
    height=600
)

fig.show()

# Interpretasi:
# - Nilai mendekati 1 (merah): Korelasi positif kuat
# - Nilai mendekati -1 (biru): Korelasi negatif kuat
# - Nilai mendekati 0 (putih): Tidak ada korelasi
# Contoh: petal_length & petal_width berkorelasi sangat kuat (0.96)
</VSCode.Cell>

<VSCode.Cell language="markdown">
### 3.4 Distribution Plot dengan Multiple Variables

Membandingkan distribusi dari multiple variabel atau groups.
</VSCode.Cell>

<VSCode.Cell language="python">
# Histogram dengan multiple variables untuk perbandingan
# Membuat data dalam format long untuk visualisasi lebih mudah
iris_long = df_iris.melt(
    id_vars=['species'],
    value_vars=['sepal_length', 'sepal_width', 'petal_length', 'petal_width'],
    var_name='measurement',
    value_name='value'
)

fig = px.histogram(
    iris_long,
    x='value',
    color='measurement',
    facet_col='measurement',  # Buat subplot untuk setiap measurement
    facet_col_wrap=2,  # 2 kolom
    title='Distribusi Semua Measurements - Iris Dataset',
    labels={'value': 'Value (cm)', 'measurement': 'Measurement Type'},
    nbins=20
)

fig.update_yaxes(matches=None)  # Y-axis independent untuk setiap subplot
fig.show()

# Penjelasan: Facet plots memudahkan perbandingan distribusi multiple variabel
# Kita bisa melihat range dan shape distribusi setiap measurement
</VSCode.Cell>

<VSCode.Cell language="markdown">
### 3.5 Contour Plot

Contour plot menunjukkan data 3D dalam 2D menggunakan kontur (garis level).

**Kapan menggunakan**: Untuk visualisasi fungsi 2D, density plots, atau data spasial.
</VSCode.Cell>

<VSCode.Cell language="python">
# Contour Plot - Density dari 2 variabel
# Buat data dengan distribusi normal bivariat
np.random.seed(42)
n = 500

x = np.random.randn(n)
y = 2 * x + np.random.randn(n)

fig = go.Figure(data=go.Histogram2dContour(
    x=x,
    y=y,
    colorscale='Blues',
    showscale=True,
    contours=dict(
        coloring='heatmap',
        showlabels=True
    )
))

fig.update_layout(
    title='Contour Plot: Density Distribution',
    xaxis_title='Variable X',
    yaxis_title='Variable Y',
    width=700,
    height=600
)

fig.show()

# Penjelasan: Contour plot menunjukkan density (kerapatan) data
# Area dengan warna lebih gelap = density lebih tinggi
# Garis kontur menghubungkan titik-titik dengan density sama
</VSCode.Cell>

<VSCode.Cell language="markdown">
### 3.6 3D Scatter Plot

Scatter plot 3 dimensi untuk visualisasi data dengan 3 variabel numerik.

**Kapan menggunakan**: Analisis data 3D, clustering visualization, multivariate analysis.
</VSCode.Cell>

<VSCode.Cell language="python">
# 3D Scatter Plot - Iris dataset
fig = px.scatter_3d(
    df_iris,
    x='sepal_length',
    y='sepal_width',
    z='petal_length',
    color='species',
    size='petal_width',
    title='3D Scatter Plot: Iris Dataset',
    labels={
        'sepal_length': 'Sepal Length (cm)',
        'sepal_width': 'Sepal Width (cm)',
        'petal_length': 'Petal Length (cm)',
        'species': 'Species'
    },
    size_max=15,
    opacity=0.7
)

# Customization camera angle
fig.update_layout(
    scene=dict(
        xaxis_title='Sepal Length',
        yaxis_title='Sepal Width',
        zaxis_title='Petal Length',
    )
)

fig.show()

# Fitur Interaktif 3D:
# - Rotate: Click dan drag untuk memutar
# - Zoom: Scroll atau pinch
# - Pan: Shift + click dan drag
# Insight: 3D plot menunjukkan pemisahan cluster spesies lebih jelas
</VSCode.Cell>

<VSCode.Cell language="markdown">
### 3.7 Density Heatmap (2D Histogram)

2D Histogram atau density heatmap untuk melihat distribusi 2 variabel.
</VSCode.Cell>

<VSCode.Cell language="python">
# Density Heatmap - 2D Histogram
fig = px.density_heatmap(
    df_tips,
    x='total_bill',
    y='tip',
    marginal_x='histogram',  # Histogram di margin X
    marginal_y='histogram',  # Histogram di margin Y
    title='Density Heatmap: Total Bill vs Tip',
    labels={'total_bill': 'Total Bill ($)', 'tip': 'Tip ($)'},
    color_continuous_scale='Viridis'
)

fig.show()

# Penjelasan: 
# - Density heatmap menunjukkan konsentrasi data di area tertentu
# - Marginal plots menunjukkan distribusi masing-masing variabel
# - Warna lebih terang = density lebih tinggi
# Insight: Kebanyakan data terkonsentrasi di bill $10-$25 dengan tip $2-$4
</VSCode.Cell>

<VSCode.Cell language="markdown">
---
## 4. Dash - Framework Web untuk Plotly

**Dash** adalah framework Python untuk membangun web application dengan visualisasi data interaktif. Dash menggunakan Plotly untuk grafik dan React.js untuk UI, tanpa perlu menulis JavaScript.

**Kegunaan Dash:**
- Membuat dashboard interaktif
- Web application untuk data analytics
- Real-time monitoring systems
- Report dan presentation tools
</VSCode.Cell>

<VSCode.Cell language="markdown">
### 4.1 Instalasi Dash

Dash terdiri dari beberapa komponen:
- **dash**: Core framework
- **dash-core-components**: Komponen interaktif (grafik, dropdown, slider, dll)
- **dash-html-components**: Komponen HTML
- **dash-bootstrap-components**: Bootstrap styling (opsional)
</VSCode.Cell>

<VSCode.Cell language="python">
# Instalasi Dash
# Jalankan perintah ini di terminal:
# pip install dash
# pip install dash-bootstrap-components  # (opsional, untuk styling)

# Import Dash libraries (gunakan jika sudah terinstall)
# import dash
# from dash import dcc, html, Input, Output
# import dash_bootstrap_components as dbc

print("Untuk menggunakan Dash, install terlebih dahulu:")
print("pip install dash dash-bootstrap-components")
</VSCode.Cell>

<VSCode.Cell language="markdown">
### 4.2 Struktur Dasar Dash App

Dash app terdiri dari 2 komponen utama:
1. **Layout**: Struktur visual app (HTML + components)
2. **Callbacks**: Interaktivitas (menghubungkan input dan output)

**Catatan**: Contoh Dash di bawah ini adalah kode untuk file Python terpisah (.py), karena Dash memerlukan server web untuk berjalan.
</VSCode.Cell>

<VSCode.Cell language="markdown">
### 4.3 Contoh 1: Dash App Sederhana

Simpan kode berikut dalam file `app_simple.py` dan jalankan dengan: `python app_simple.py`

```python
import dash
from dash import dcc, html
import plotly.express as px
import pandas as pd

# Inisialisasi Dash app
app = dash.Dash(__name__)

# Load data
df = px.data.gapminder()

# Buat grafik
fig = px.scatter(
    df[df['year'] == 2007],
    x='gdpPercap',
    y='lifeExp',
    size='pop',
    color='continent',
    hover_name='country',
    title='Life Expectancy vs GDP (2007)',
    log_x=True
)

# Layout app
app.layout = html.Div([
    html.H1('Dashboard Sederhana dengan Dash'),
    html.P('Ini adalah contoh dashboard sederhana menggunakan Dash dan Plotly'),
    dcc.Graph(
        id='scatter-plot',
        figure=fig
    )
])

# Jalankan server
if __name__ == '__main__':
    app.run_server(debug=True)
```

**Penjelasan:**
- `app = dash.Dash(__name__)`: Inisialisasi aplikasi Dash
- `app.layout`: Mendefinisikan struktur visual (HTML)
- `html.H1`, `html.P`: Komponen HTML
- `dcc.Graph`: Komponen untuk menampilkan grafik Plotly
- `app.run_server()`: Menjalankan web server (akses di http://127.0.0.1:8050/)
</VSCode.Cell>

<VSCode.Cell language="markdown">
### 4.4 Contoh 2: Dash App dengan Interaktivity (Callbacks)

Callback membuat app interaktif dengan menghubungkan input (dropdown, slider, dll) ke output (grafik, text, dll).

**Struktur Callback:**
```python
@app.callback(
    Output(component_id, component_property),
    Input(component_id, component_property)
)
def update_function(input_value):
    # Process input
    # Return output
    return output_value
```
</VSCode.Cell>

<VSCode.Cell language="markdown">
**Contoh Lengkap dengan Callback** - Simpan dalam `app_interactive.py`:

```python
import dash
from dash import dcc, html, Input, Output
import plotly.express as px
import pandas as pd

# Inisialisasi app
app = dash.Dash(__name__)

# Load data
df = px.data.gapminder()

# Layout
app.layout = html.Div([
    html.H1('Dashboard Interaktif - Gapminder Dataset'),
    
    html.Div([
        html.Label('Pilih Tahun:'),
        dcc.Slider(
            id='year-slider',
            min=df['year'].min(),
            max=df['year'].max(),
            step=5,
            value=2007,
            marks={str(year): str(year) for year in df['year'].unique()}
        )
    ]),
    
    html.Br(),
    
    html.Div([
        html.Label('Pilih Benua:'),
        dcc.Dropdown(
            id='continent-dropdown',
            options=[{'label': cont, 'value': cont} for cont in df['continent'].unique()] + 
                    [{'label': 'All', 'value': 'All'}],
            value='All',
            clearable=False
        )
    ], style={'width': '50%'}),
    
    html.Br(),
    
    dcc.Graph(id='scatter-plot')
])

# Callback untuk update grafik
@app.callback(
    Output('scatter-plot', 'figure'),
    [Input('year-slider', 'value'),
     Input('continent-dropdown', 'value')]
)
def update_graph(selected_year, selected_continent):
    # Filter data berdasarkan tahun
    filtered_df = df[df['year'] == selected_year]
    
    # Filter berdasarkan benua jika bukan 'All'
    if selected_continent != 'All':
        filtered_df = filtered_df[filtered_df['continent'] == selected_continent]
    
    # Buat grafik
    fig = px.scatter(
        filtered_df,
        x='gdpPercap',
        y='lifeExp',
        size='pop',
        color='continent',
        hover_name='country',
        title=f'Life Expectancy vs GDP ({selected_year})',
        log_x=True,
        size_max=60
    )
    
    fig.update_layout(transition_duration=500)  # Animasi transisi
    
    return fig

if __name__ == '__main__':
    app.run_server(debug=True)
```

**Penjelasan Callback:**
- `@app.callback`: Decorator untuk mendefinisikan callback
- `Output('scatter-plot', 'figure')`: Output adalah properti 'figure' dari komponen 'scatter-plot'
- `Input('year-slider', 'value')`: Input dari nilai slider tahun
- `Input('continent-dropdown', 'value')`: Input dari dropdown benua
- `update_graph()`: Fungsi yang dipanggil saat input berubah
- Return value dari fungsi menjadi output (update grafik)

**Komponen Interaktif yang Digunakan:**
1. `dcc.Slider`: Untuk memilih tahun (range slider)
2. `dcc.Dropdown`: Untuk memilih benua
3. Grafik otomatis update saat user mengubah slider atau dropdown
</VSCode.Cell>

<VSCode.Cell language="markdown">
### 4.5 Contoh 3: Dashboard Complex dengan Multiple Charts

Dashboard yang lebih kompleks dengan multiple charts, filters, dan styling Bootstrap.

Simpan dalam file `app_dashboard.py`:

```python
import dash
from dash import dcc, html, Input, Output
import dash_bootstrap_components as dbc
import plotly.express as px
import plotly.graph_objects as go
import pandas as pd

# Inisialisasi app dengan Bootstrap theme
app = dash.Dash(__name__, external_stylesheets=[dbc.themes.BOOTSTRAP])

# Load data
df_tips = px.data.tips()

# Layout dengan Bootstrap
app.layout = dbc.Container([
    dbc.Row([
        dbc.Col([
            html.H1('Restaurant Tips Dashboard', className='text-center mb-4'),
            html.Hr()
        ])
    ]),
    
    dbc.Row([
        dbc.Col([
            html.Label('Filter by Day:'),
            dcc.Dropdown(
                id='day-filter',
                options=[{'label': 'All Days', 'value': 'All'}] +
                        [{'label': day, 'value': day} for day in df_tips['day'].unique()],
                value='All',
                clearable=False
            )
        ], width=6),
        
        dbc.Col([
            html.Label('Filter by Time:'),
            dcc.Dropdown(
                id='time-filter',
                options=[{'label': 'All', 'value': 'All'}] +
                        [{'label': time, 'value': time} for time in df_tips['time'].unique()],
                value='All',
                clearable=False
            )
        ], width=6)
    ], className='mb-4'),
    
    dbc.Row([
        dbc.Col([
            dcc.Graph(id='scatter-chart')
        ], width=6),
        
        dbc.Col([
            dcc.Graph(id='box-chart')
        ], width=6)
    ], className='mb-4'),
    
    dbc.Row([
        dbc.Col([
            dcc.Graph(id='bar-chart')
        ], width=6),
        
        dbc.Col([
            dcc.Graph(id='pie-chart')
        ], width=6)
    ])
], fluid=True)

# Callback untuk update semua charts
@app.callback(
    [Output('scatter-chart', 'figure'),
     Output('box-chart', 'figure'),
     Output('bar-chart', 'figure'),
     Output('pie-chart', 'figure')],
    [Input('day-filter', 'value'),
     Input('time-filter', 'value')]
)
def update_charts(selected_day, selected_time):
    # Filter data
    filtered_df = df_tips.copy()
    
    if selected_day != 'All':
        filtered_df = filtered_df[filtered_df['day'] == selected_day]
    
    if selected_time != 'All':
        filtered_df = filtered_df[filtered_df['time'] == selected_time]
    
    # Chart 1: Scatter plot
    fig_scatter = px.scatter(
        filtered_df,
        x='total_bill',
        y='tip',
        color='sex',
        size='size',
        title='Total Bill vs Tip',
        labels={'total_bill': 'Total Bill ($)', 'tip': 'Tip ($)'}
    )
    
    # Chart 2: Box plot
    fig_box = px.box(
        filtered_df,
        x='day',
        y='tip',
        color='sex',
        title='Tip Distribution by Day'
    )
    
    # Chart 3: Bar chart
    avg_tips = filtered_df.groupby('day')['tip'].mean().reset_index()
    fig_bar = px.bar(
        avg_tips,
        x='day',
        y='tip',
        title='Average Tip by Day',
        labels={'tip': 'Average Tip ($)'}
    )
    
    # Chart 4: Pie chart
    tips_by_time = filtered_df.groupby('time')['tip'].sum().reset_index()
    fig_pie = px.pie(
        tips_by_time,
        values='tip',
        names='time',
        title='Tip Distribution by Time'
    )
    
    return fig_scatter, fig_box, fig_bar, fig_pie

if __name__ == '__main__':
    app.run_server(debug=True, port=8050)
```

**Fitur Dashboard ini:**
- 2 Dropdowns untuk filtering (hari dan waktu)
- 4 Charts yang saling terhubung (scatter, box, bar, pie)
- Bootstrap layout untuk tampilan responsive dan rapi
- Real-time updates - semua charts update simultaneously saat filter berubah
- Professional styling dengan Bootstrap theme
</VSCode.Cell>

<VSCode.Cell language="markdown">
### 4.6 Tips dan Best Practices Dash

**1. Layout Organization:**
- Gunakan `dbc.Container`, `dbc.Row`, `dbc.Col` untuk layout responsive
- Pisahkan styling dengan CSS external atau inline styles
- Gunakan `className` untuk Bootstrap classes (e.g., 'text-center', 'mb-4')

**2. Callbacks Optimization:**
- Batasi jumlah callbacks untuk performa optimal
- Gunakan `prevent_initial_call=True` jika tidak perlu render awal
- Combine multiple outputs dalam satu callback (lebih efisien)
- Gunakan `dash.callback_context` untuk handle multiple inputs
- Consider using `@cache.memoize` untuk cache hasil komputasi berat

**3. Performance Tips:**
- Cache data dengan `@cache.memoize` untuk dataset besar
- Gunakan `dcc.Loading` untuk loading indicators saat processing
- Optimize data processing dalam callbacks
- Gunakan `clientside_callback` untuk operasi sederhana (lebih cepat)

**4. Komponen Dash yang Sering Digunakan:**

**Core Components (dcc):**
- `dcc.Graph`: Untuk Plotly charts
- `dcc.Dropdown`: Dropdown menu
- `dcc.Slider`: Slider untuk range
- `dcc.RangeSlider`: Slider untuk range dengan 2 handle
- `dcc.DatePickerRange`: Date picker
- `dcc.Input`: Text input
- `dcc.Textarea`: Multi-line text
- `dcc.Checklist`: Checkboxes
- `dcc.RadioItems`: Radio buttons
- `dcc.Loading`: Loading spinner

**HTML Components (html):**
- `html.Div`: Container div
- `html.H1`, `html.H2`, dll: Headers
- `html.P`: Paragraph
- `html.Br`: Line break
- `html.Hr`: Horizontal rule
- `html.Button`: Button

**5. Deployment:**
- **Local**: `python app.py` (untuk development)
- **Production Options**:
  - Heroku (gratis untuk hobby projects)
  - PythonAnywhere
  - AWS/Google Cloud/Azure
  - Dash Enterprise (komersial, full-featured)

**6. Resources Tambahan:**
- Documentation: https://dash.plotly.com/
- Community Forum: https://community.plotly.com/
- Examples Gallery: https://dash-gallery.plotly.host/Portal/
</VSCode.Cell>

<VSCode.Cell language="markdown">
---
## Kesimpulan

Tutorial ini telah membahas:

### 1. **Instalasi dan Penggunaan Sederhana**
- Cara install Plotly dan dependencies
- Import library
- Membuat grafik pertama (scatter plot sederhana)
- Fitur interaktif Plotly

### 2. **Basic Charts**
- Scatter Plot: Hubungan antar variabel
- Line Chart: Trend dan time series
- Bar Chart: Perbandingan kategori (vertical, horizontal, grouped)
- Pie Chart & Donut Chart: Proporsi dan komposisi
- Histogram: Distribusi frekuensi
- Box Plot: Distribusi dengan quartiles

### 3. **Statistical Charts**
- Violin Plot: Distribusi detail dengan density
- Scatter Matrix: Multiple variable relationships
- Correlation Heatmap: Korelasi antar variabel
- Distribution Plot: Multiple variable distributions
- Contour Plot: Density dalam 2D
- 3D Scatter Plot: Visualisasi 3 dimensi
- Density Heatmap: 2D histogram dengan marginals

### 4. **Dash Framework**
- Instalasi dan setup Dash
- Struktur dasar Dash app (Layout)
- Callbacks untuk interactivity
- Multiple charts dashboard
- Best practices dan tips optimization

**Next Steps:**
1. Practice dengan dataset Anda sendiri
2. Explore dokumentasi Plotly untuk fitur lanjutan
3. Buat dashboard interaktif dengan Dash
4. Pelajari advanced features: animations, subplots, custom styling
5. Deploy dashboard Anda ke production

**Resources:**
- Plotly Documentation: https://plotly.com/python/
- Dash Documentation: https://dash.plotly.com/
- Plotly Community: https://community.plotly.com/

Selamat bereksperimen dengan Plotly! 🎉📊
</VSCode.Cell>
````
`````

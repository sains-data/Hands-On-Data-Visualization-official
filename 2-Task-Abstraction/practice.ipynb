## Dataset dan Pustaka

```python
import plotly.express as px
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

# Memuat dataset global (Gapminder) untuk latihan
df_gapminder = px.data.gapminder()
# Memuat dataset tips (E-commerce sederhana)
df_tips = px.data.tips()
```

## Contoh Skenario 1 & 2: Mencari Tren dan Outlier (Global Health) - Data Gapminder

Domain Task: "Identifikasi negara yang menyimpang dari tren umum GDP vs Harapan Hidup.

```python
# Mengambil data tahun terakhir saja
df_2007 = df_gapminder.query("year == 2007")

# Visualisasi Scatterplot untuk Correlation & Outliers
fig = px.scatter(df_2007, x="gdpPercap", y="lifeExp", 
                 color="continent", size="pop", hover_name="country",
                 log_x=True, title="Skenario 1 & 2: Correlation & Outliers Detection")

# Menambahkan anotasi manual untuk outlier (Contoh: Kuwait yang sangat kaya tapi lifeExp tidak tertinggi)
fig.add_annotation(x=40000, y=75, text="Potensi Outlier", showarrow=True, arrowhead=1)
fig.show()
```

# Contoh Skenario 3: Membandingkan Distribusi (Ekonomi)

Domain task: Membandingkan distribusi pendapatan (GDP) di setiap benua untuk melihat ketimpangan.

```python
# Menggunakan Box Plot untuk Target: Distribution
fig2 = px.box(df_2007, x="continent", y="gdpPercap", color="continent",
              points="all", title="Skenario 3: Distribution & Comparison")
fig2.show()
```

# Contoh Skenario 8: Menghasilkan Data Baru (Produce: Derive)
Domain task : User ingin tahu persentase tip, namun data tersebut tidak ada di kolom asli. Kita harus memproduksinya.

```python
# Action: Produce -> Derive (Membuat kolom baru)
df_tips['tip_percentage'] = (df_tips['tip'] / df_tips['total_bill']) * 100

# Visualisasi data hasil Derive
fig3 = px.histogram(df_tips, 
                    x="tip_percentage", 
                    nbins=20, 
                   title="Skenario 8: Hasil dari Action 'Derive' (Persentase Tip)")
fig3.show()
```
